## Handle Missing Values

```python
from sklearn.compose import ColumnTransformer
df = pd.read_csv('../data/car-sales-extended-missing-data.csv')
```

```python
df.head()
```

```python
# Cek Missing Values
df.isna().sum()
```

```python
# Hapus kolom target yang ada missing values
df.dropna(subset='Price', inplace=True)
```

```python
df.isna().sum()
```

```python
# Data Splitting
X = df.drop('Price', axis=1)
y = df['Price']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)
```

```python
X_train.head()
```

```python
# Handle Missing Values
from sklearn.impute import SimpleImputer
# apa yang bakal dilakuin untuk tiap jenis kolom
cat_imputer = SimpleImputer(strategy='constant', fill_value='missing') # Category = 'missing'
num_imputer = SimpleImputer(strategy='mean') # Numerik = nilai mean kolom
door_imputer = SimpleImputer(strategy='constant', fill_value=4) # Kolom Doors = 4 pintu di mobil umumnya 4

# misahin kolom yang bakal di-impute
cat_feat = ['Make', 'Colour']
num_feat = ['Odometer (KM)']
door_feat = ['Doors']

# gabung semua imputer
imputer = ColumnTransformer([
    ('cat_imputer', cat_imputer, cat_feat),
    ('num_imputer', num_imputer, num_feat),
    ('door_imputer', door_imputer, door_feat)
])

# fill missing values
filled_X_train = imputer.fit_transform(X_train)
filled_X_test = imputer.transform(X_test)
```

```python
# ubah supaya jadi DataFrame
x_train = pd.DataFrame(filled_X_train, columns=X_train.columns)
x_test = pd.DataFrame(filled_X_test, columns=X_test.columns)
```

## Feature Encoding

```python
# Opsi 1: encoding doang
from sklearn.preprocessing import OneHotEncoder
# pilih kolom yang mau di-encoding
categorical_features = ['Make', 'Colour', 'Doors']
# membuat transformer
one_hot = OneHotEncoder()
transformer = ColumnTransformer(
    [('one_hot', one_hot, categorical_features)],
    remainder='passthrough'
)
# encoding
X_train_encoded = transformer.fit_transform(x_train)
X_test_encoded = transformer.transform(x_test)
```
