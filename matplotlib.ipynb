## Matplotlib Workflow

```python
# 0. import matplotlib liblary
import matplotlib.pyplot as plt
# 1. prepare data
x = [1,2,3,4,5]
y = [11,22,33,44,55]
# 2. setup plot
fig, ax = plt.subplots(figsize=(5,5)) # figsize = (widht, height)
# 3. plot data
ax.plot(x,y)
# 4. customize plot
ax.set(
    title='Simple Plot',
    xlabel='X',
    ylabel='Y'
)
# 5. save & show 
fig.savefig('img/sample-plot.png')
```

```python
nilai_uas = {'Matematika': 75, 'B.Inggris': 97, 'Fisika': 100, 'Kimia': 80, 'Agama': 100, 'Sejarah': 70, 'PKN': 50}
```

```python
# bar plot
fig, ax = plt.subplots()
ax.bar(nilai_uas.keys(), nilai_uas.values())
ax.set(
    title='Nilai UAS Azharaul',
    ylabel='Nilai',
    xlabel='Mata Pelajaran'
);
fig.tight_layout()
```

```python
# horizontal bar plot
fig, ax = plt.subplots()
ax.barh(nilai_uas.keys(), nilai_uas.values())
ax.set(
    title='Nilai UAS Azharaul',
    ylabel='Mata Pelajaranilai',
    xlabel='Nilai'
);
```

```python
# histogram
x = np.random.randn(1000)
fig, ax = plt.subplots()
ax.hist(x);
```

```python
# subplots opsi 1
fig, ((ax1, ax2),(ax3, ax4)) = plt.subplots(nrows=2, ncols=2, figsize=(6,6))
ax1.plot(x, x/2)
ax1.set(title="Plot", xlabel="X", ylabel='Y')

ax2.scatter(np.random.random(10), np.random.random(10))
ax2.set(title="Scatter", xlabel="X", ylabel='Y')

ax3.barh(nilai_uas.keys(), nilai_uas.values())
ax3.set(title="Horizontal Bar", xlabel="Nilai", ylabel='Mata Pelajaran')

ax4.hist(x);
ax4.set(title="Histogram")

fig.tight_layout()
```

```python
# subplots opsi 2
fig, ax = plt.subplots(nrows=2, ncols=2, figsize=(10,5))
ax[0,0].plot(x, x/2)
ax[0,0].set(title="Plot", xlabel="X", ylabel='Y')

ax[0,1].scatter(np.random.random(10), np.random.random(10))
ax[0,1].set(title="Scatter", xlabel="X", ylabel='Y')

ax[1,0].barh(nilai_uas.keys(), nilai_uas.values())
ax[1,0].set(title="Horizontal Bar", xlabel="Nilai", ylabel='Mata Pelajaran')

ax[1,1].hist(x);
ax[1,1].set(title="Histogram")

fig.tight_layout()
```

## Plotting dari DataFrame

```python
df = pd.read_csv('data/car-sales.csv')
df.head()
```

```python
df['Price'] = df['Price'].str.replace(r'[\$\,\.]', '', regex=True)
df
```

```python
df['Price'] = df['Price'].str[:-2]
```

```python
df['Price'] = df['Price'].astype(int)
```

```python
df['Sale Date'] = pd.date_range('1/1/2026', periods=len(df))
df
```

```python
df['Total Sales'] = df['Price'].cumsum()
df
```

```python
df.plot(x='Sale Date', y='Total Sales');
```

```python
df.plot(x='Odometer (KM)', y='Price', kind='scatter');
```

```python
df.plot(kind='box')
```
