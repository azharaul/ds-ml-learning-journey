```python
# Variance (Varians) = Ukuran seberapa jauh penyebaran angka-angka atau titik data dari nilai rata-ratanya (mean).
np.var(a2)
```

```python
# Standard Deviation (Simpangan Baku) = Akar kuadrat dari variance.
np.std(a2)
```

```python
# contoh akar kuadrat dari variance
np.sqrt(np.var(a2))
```

```python
high_var_array = np.array([1, 100, 200, 300, 4000, 5000])
low_var_array = np.array([2, 4, 6, 8, 10])
```

```python
print("Variance")
print(f'High Var Array: {np.var(high_var_array)}')
print(f'Low Var Array: {np.var(low_var_array)}')
```

```python
print("Standart Deviasi")
print(f'High Var Array: {np.std(high_var_array)}')
print(f'Low Var Array: {np.std(low_var_array)}')
```

```python
print("Rata Rata")
print(f'High Var Array: {np.mean(high_var_array)}')
print(f'Low Var Array: {np.mean(low_var_array)}')
```

```python
import matplotlib.pyplot as plt
plt.hist(high_var_array)
plt.show()
```

```python
plt.hist(low_var_array)
plt.show()
```

## Dot product example (nut butter sales)

```python
np.random.seed(2)
a_index = "Senin Selasa Rabu Kamis Jumat".split()
a_col = "Mouse Keyboard Monitor".split()
a_val = np.random.randint(20, size=(5, 3))
```

```python
data_alat = pd.DataFrame(a_val, columns=a_col, index=a_index)
data_alat
```

```python
harga_alat = np.array([[100000, 500000, 1200000]])
daftar_harga = pd.DataFrame(harga_alat, columns=a_col, index=['Harga'])
daftar_harga
```

```python
total = np.dot(data_alat, daftar_harga.T)
data_alat['Total (Rp)'] = total
```

```python
data_alat
```

```python
daftar_harga
```
