## Matplotlib Workflow

```python
# 0. import matplotlib liblary
import matplotlib.pyplot as plt
# 1. prepare data
x = [1,2,3,4,5]
y = [11,22,33,44,55]
# 2. setup plot
fig, ax = plt.subplots(figsize=(5,5)) # figsize = (widht, height)
# 3. plot data
ax.plot(x,y)
# 4. customize plot
ax.set(
    title='Simple Plot',
    xlabel='X',
    ylabel='Y'
)
# 5. save & show 
fig.savefig('img/sample-plot.png')
```

```python
nilai_uas = {'Matematika': 75, 'B.Inggris': 97, 'Fisika': 100, 'Kimia': 80, 'Agama': 100, 'Sejarah': 70, 'PKN': 50}
```

```python
# bar plot
fig, ax = plt.subplots()
ax.bar(nilai_uas.keys(), nilai_uas.values())
ax.set(
    title='Nilai UAS Azharaul',
    ylabel='Nilai',
    xlabel='Mata Pelajaran'
);
fig.tight_layout()
```

```python
# horizontal bar plot
fig, ax = plt.subplots()
ax.barh(nilai_uas.keys(), nilai_uas.values())
ax.set(
    title='Nilai UAS Azharaul',
    ylabel='Mata Pelajaranilai',
    xlabel='Nilai'
);
```

```python
# histogram
x = np.random.randn(1000)
fig, ax = plt.subplots()
ax.hist(x);
```

```python
# subplots opsi 1
fig, ((ax1, ax2),(ax3, ax4)) = plt.subplots(nrows=2, ncols=2, figsize=(6,6))
ax1.plot(x, x/2)
ax1.set(title="Plot", xlabel="X", ylabel='Y')

ax2.scatter(np.random.random(10), np.random.random(10))
ax2.set(title="Scatter", xlabel="X", ylabel='Y')

ax3.barh(nilai_uas.keys(), nilai_uas.values())
ax3.set(title="Horizontal Bar", xlabel="Nilai", ylabel='Mata Pelajaran')

ax4.hist(x);
ax4.set(title="Histogram")

fig.tight_layout()
```

```python
# subplots opsi 2
fig, ax = plt.subplots(nrows=2, ncols=2, figsize=(10,5))
ax[0,0].plot(x, x/2)
ax[0,0].set(title="Plot", xlabel="X", ylabel='Y')

ax[0,1].scatter(np.random.random(10), np.random.random(10))
ax[0,1].set(title="Scatter", xlabel="X", ylabel='Y')

ax[1,0].barh(nilai_uas.keys(), nilai_uas.values())
ax[1,0].set(title="Horizontal Bar", xlabel="Nilai", ylabel='Mata Pelajaran')

ax[1,1].hist(x);
ax[1,1].set(title="Histogram")

fig.tight_layout()
```

### Metode apa yang bagus buat dipake? (pyplot atau oo matplotlib)
* pake pyplot kalo mau bikin plot cepet
* kalo mau plotnya lebih advance/bagus pake object oriented matplotlib soalnya bisa lebih bebas buat di kostumisasi

```python
over_50 = heart_disease[heart_disease['age']>50]
over_50.tail()
```

```python
# pylot method
over_50.plot(kind='scatter', x='age', y='chol', c='target', figsize=(10,6));
```

```python
# OO method mixed with pyplot method
fig, ax = plt.subplots(figsize=(10,6))
over_50.plot(kind='scatter', x='age', y='chol',c='target', ax=ax);
ax.set_xlim([45, 100])
```

```python
# OO method murni
fig, ax = plt.subplots(figsize=(10,6))

scatter = ax.scatter(x=over_50['age'], y=over_50['chol'], c=over_50['target'])

# kostumisasi plot
ax.set(title='Penyakit jantung dan Level Kolesterol', xlabel='Umur', ylabel='Kolesterol')
ax.legend(*scatter.legend_elements(), title='Target')

# tambah garis horizontal
ax.axhline(over_50['chol'].mean(), linestyle='--')

# tampilkan plot
plt.show()
```

```python
fig, (ax0, ax1) = plt.subplots(figsize=(10, 10), nrows=2, ncols=1, sharex=True)

scatter = ax0.scatter(x=over_50['age'], y=over_50['chol'], c=over_50['target'])

ax0.set(title='Visualisasi Penyakit Jantung berdasarkan tingkat kolesterol', ylabel='Tingkat Kolesterol')
ax0.legend(*scatter.legend_elements(), title='Target')
ax0.axhline(over_50['chol'].mean(), linestyle='-.')

scatter = ax1.scatter(x=over_50['age'], y=over_50['thalach'], c=over_50['target'])
ax1.set(title='Visualisasi Penyakit Jantung berdasarkan detak jantung maksimum', xlabel='Umur', ylabel='Detak Jantung Maksumum')
ax1.legend(*scatter.legend_elements(), title='Target')
ax1.axhline(over_50['thalach'].mean(), linestyle='--')

fig.suptitle('Analisa Penyakit Jantung', fontsize=16, fontweight='bold')

plt.tight_layout()
plt.show()
```
