```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder

from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import train_test_split, GridSearchCV

from sklearn.metrics import mean_absolute_error
from sklearn.metrics import root_mean_squared_error

np.random.seed(42)
df = pd.read_csv('../data/car-sales-extended-missing-data.csv')
```

1. Fill Missing data
2. Encode Data

```python
# Drop missing data on target column
df.dropna(subset='Price', inplace=True)
```

```python
# Creating pipeline
cat_feat = ['Make', 'Colour']
cat_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='constant', fill_value='missing')),
    ('onehot', OneHotEncoder(handle_unknown='ignore'))
])
door_feat = ['Doors']
door_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='constant', fill_value=4))
])
num_feat = ['Odometer (KM)']
num_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='mean'))
])
```

Data preprocessing

```python
preprocessor = ColumnTransformer(
    transformers=[
        ('cat', cat_transformer, cat_feat),
        ('door', door_transformer, door_feat),
        ('num', num_transformer, num_feat)
    ]
)

model = Pipeline(steps=[('preprocessor', preprocessor),
                        ('model', RandomForestRegressor())])
```

```python
# Data splitting
X = df.drop('Price', axis=1)
y = df['Price']

# train test split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)
```

```python
model.fit(X_train, y_train)
model.score(X_test, y_test)
```

Hyperparameter

```python
pipe_grid = {
    'preprocessor__num__imputer__strategy':['mean', 'median'],
    'model__n_estimators':[100,1000],
    'model__max_depth': [None, 5],
    'model__max_features': ['sqrt'],
    'model__min_samples_split': [2,4]
}

gs_model = GridSearchCV(model, pipe_grid, cv=5, verbose=2)
gs_model.fit(X_train, y_train)
```

```python
gs_model.score(X_test, y_test)
```

```python
y_preds = gs_model.predict(X_test)
```

```python
mean_absolute_error(y_test, y_preds)
```

```python
root_mean_squared_error(y_test, y_preds)
```
