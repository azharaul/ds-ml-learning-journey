## End-to-end Skicit-Learn Workflow

```python
import pandas as pd
import numpy as np
heart_disease = pd.read_csv('data/heart-disease.csv')
```

```python
heart_disease.head()
```

```python
# train test split 

# x = features
X = heart_disease.drop('target', axis=1)

# y = target
y = heart_disease['target']
```

```python
# pilih model dan hyperparameters
# karena targetnya hanya punya 2 nilai, yaitu 0 dan 1, berarti model yang akan digunakan ada model klasifikasi
from sklearn.ensemble import RandomForestClassifier
clf = RandomForestClassifier() # default hyperparameters
```

```python
# kasih data ke model (fitting)
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y,  test_size=0.2) # 80% buat training, 20% buat test
```

```python
clf.fit(X_train, y_train)
```

```python
# prediksi
y_preds = clf.predict(X_test)
```

```python
# model evaluasi
clf.score(X_train, y_train)
```

```python
clf.score(X_test, y_test)
```

```python
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score

print(classification_report(y_test, y_preds))
```

```python
confusion_matrix(y_test, y_preds)
```

```python
accuracy_score(y_test, y_preds)
```

```python
# Improve model
np.random.seed(42)
for i in range(10, 100, 10):
    print(f'Modelling menggunakan {i} estimators...')
    clf = RandomForestClassifier(n_estimators=i).fit(X_train, y_train)
    print(f"Akurasi model saat test: {clf.score(X_test, y_test) * 100:.2f}%")
    print("")
```

```python
# save model
import pickle
pickle.dump(clf, open('random_forest_model_1.pkl', 'wb'))
```

```python
# load model
loaded_model = pickle.load(open('random_forest_model_1.pkl', 'rb'))
loaded_model.score(X_test, y_test)
```

### Memastikan bahwa semua data yang akan digunakan ber-tipe numerik

```python
car_sales = pd.read_csv('data/scikit-learn-data/car-sales-extended.csv')
```

```python
car_sales.head()
```

di kolom Make dan Colour bukan numerik, oleh karena itu kita encoding dulu supaya jadi numerik

```python
# data splitting
# from sklearn.model_selection import train_test_split

X = car_sales.drop('Price', axis=1)
y = car_sales['Price']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)
```

```python
# Membuat model machine learning (ini bakal error soalnya belum di-encoding)
from sklearn.ensemble import RandomForestRegressor # karena kali ini target-nya berupa numerik kontinu jadi harus pake model regresi

model = RandomForestRegressor()
model.fit(X_train, y_train)
model.score(X_test, y_test)
```

```python
# Ubah data kategorikal ke numerikal
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer

categorical_features = ['Make', 'Colour', 'Doors']
one_hot = OneHotEncoder()
transformer = ColumnTransformer([('one_hot', one_hot, categorical_features)], remainder='passthrough')
transformed_X = transformer.fit_transform(X)
```

```python
X.head() # sebelum di-encoding
```

```python
pd.DataFrame(transformed_X).head() # setelah di-encoding
```

```python
np.random.seed(42)
# split pake data yang udah di encoding
X_train, X_test, y_train, y_test = train_test_split(transformed_X, y, train_size=0.2)
model.fit(X_train, y_train)
```

```python
model.score(X_test, y_test)
```

### Kalo datanya ada missing values
1. Imputation = Isi data yang hilang
2. Hapus data yg hilang

```python
car_sales_missing = pd.read_csv('data/scikit-learn-data/car-sales-extended-missing-data.csv')
car_sales_missing.head()
```

```python
# cek missing value
car_sales_missing.isna().sum()
```

#### Opsi 1: Fill Missing data pake pandas

```python
car_sales_missing['Doors'].value_counts()
```

```python
# Fill 'Make' Column
car_sales_missing['Make'] = car_sales_missing['Make'].fillna('Missing')

# Fill 'Colour' Column
car_sales_missing['Colour'] = car_sales_missing['Colour'].fillna('Missing')

# Fill 'Odometer (KM)' Column
car_sales_missing['Odometer (KM)'] = car_sales_missing['Odometer (KM)'].fillna(car_sales_missing['Odometer (KM)'].mean())

# Fill 'Doors' Column
car_sales_missing['Doors'] = car_sales_missing['Doors'].fillna(4)
```

```python
car_sales_missing.isna().sum()
```

```python
# Hapus baris yang missing valuenya di bagian target 'Price'
car_sales_missing.dropna(inplace=True)
```

```python
car_sales_missing.isna().sum()
```

```python
X = car_sales_missing.drop('Price', axis=1)
y = car_sales_missing['Price']
```

```python
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer

categorical_features = ['Make', 'Colour', 'Doors']
one_hot = OneHotEncoder()
transformer = ColumnTransformer([('one_hot', one_hot, categorical_features)], remainder='passthrough')

transformed_X = transformer.fit_transform(X)
```
