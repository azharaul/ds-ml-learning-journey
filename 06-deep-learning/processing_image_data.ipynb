```python
# Importing the image module from matplotlib liblary
import matplotlib.image as mpimg
import matplotlib.pyplot as plt
```

Resizing the image using Pillow

```python
from PIL import Image
```

```python
img = Image.open('cat')
img_resized = img.resize((200, 200))
```

```python
img_resized.save('cat_resized.jpg')
```

```python
img_resized = mpimg.imread('cat_resized.jpg')
img_resized_plt = plt.imshow(img_resized)
```

```python
img_resized.shape
```

Converting RGB images to grayscale image using OpenCV

```python
import cv2
```

```python
img = cv2.imread('cat_resized.jpg')
```

```python
type(img)
```

```python
img.shape
```

```python
grayscale_image = cv2.cvtColor(img, cv2.COLOR_RGB2GRAY)
```

```python
grayscale_image.shape
```

```python
from google.colab.patches import cv2_imshow
```

```python
cv2_imshow(grayscale_image)
```

```python
cv2.imwrite('cat_res_gray.jpg', grayscale_image)
```
