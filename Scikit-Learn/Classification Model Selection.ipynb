```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score
```

Data Import

```python
df = pd.read_csv('../data/heart-disease.csv')
```

Data Splitting


```python
X = df.drop('target', axis=1)
y = df['target']
```

Train Test Split


```python
np.random.seed(42)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)
```

Model Fitting

```python
model = RandomForestClassifier()
model.fit(X_train, y_train)
model.score(X_test, y_test)
```

##### Model Prediction
* predict()
* predict_proba() `Cuma ada di model klasifikasi`

```python
y_preds = model.predict(X_test) # Prediksi menggunakan test data
```

```python
model.predict_proba(X_test[:5])
```

Model Evaluation

```python
# membandingkan hasil prediksi model dengan data asli
model.score(X_test, y_test) # ini sama saja seperti: np.mean(y_preds == y_test)
```

```python
accuracy_score(y_test, y_preds)
```

bedanya apa antara `model.score()` dengan `accuracy_score()`?
1. `model.score()` = method bawaan tiap model, metrik yang dihasilkan juga beda tergantung jenis modelnya.
    * Regressor = R2score 
    * Clasifier = Accuracy (Jumlah prediksi benar / Jumlah total prediksi)
2. `accuracy_score()` = ini fungsi terpisah dari sklearn-nya bukan bawaan model, jadi kalo evaluasi model regressi tapi mau cek akurasinya, bisa pake fungsi ini.
