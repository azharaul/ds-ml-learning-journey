# Introduction to Matplotlib

```python
%matplotlib inline
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
plt.plot()
```

```python
plt.plot([1,2,3,4,5])
```

```python
x = [1,2,3,4]
y = [11,22,33,44]
plt.plot(x,y);
```

```python
# 1st method
fig = plt.figure()
ax = fig.add_subplot()
```

```python
# 2nd method
fig = plt.figure()
ax = fig.add_axes([1,1,1,1])
ax.plot(x,y)
plt.show()
```

```python
# 3rd method (recommended)
fig, ax = plt.subplots()
ax.plot(x,[50,100,200,250]);
```
